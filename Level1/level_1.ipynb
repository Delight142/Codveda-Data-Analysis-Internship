```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# HISTOGRAM
df["Total day minutes"].hist(bins=20)
plt.title("Distribution of Total Day Minutes")
plt.xlabel("Total day minutes")
plt.ylabel("Number of customers")
#plt.savefig("histogram_day_minutes.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# BOX PLOT
sns.boxplot(x="Churn", y="Total day minutes", data=df)
plt.title("Total Daily Minutes: Churn vs Non-Churn")
plt.xlabel("Churn")
plt.ylabel("Total daily minutes")
#plt.savefig("boxplot_churn.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# SCATTER PLOT
sns.scatterplot(x="Total day minutes", y="Total day charge", hue="Churn", data=df)
plt.title("Total Daily Minutes vs Total Daily Charge")
plt.xlabel("Total daily minutes")
plt.ylabel("Total daily charge")
#plt.savefig("scatter_day_minutes_charge.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# CORRELATIONS
corr = df.corr(numeric_only=True).round(2)
corr
```

```python
plt.figure(figsize=(16,12))
sns.heatmap(corr, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
#plt.savefig("correlation_heatmap.png", dpi=300, bbox_inches="tight")
plt.show()
```

# LEVEL 1 TASK 3: BASIC DATA VISUALIZATION
**Dataset** churn-bigml-80

```python
df_big = pd.read_csv(r"C:\Users\USER\Downloads\Data Set For Task-20261027T131359Z-1-001\Data Set For Task\Churn Prdiction Data\churn-bigml-80.csv")
print(df_big.shape)
df_big.head()
```

```python
avg_calls = df_big.groupby("Churn")["Customer service calls"].mean().round(2)
print(avg_calls)
```

```python
sns.barplot(x="Churn", y="Customer service calls", hue="Churn", data=df_big, errorbar=None)
plt.title("Average Customer Service Calls by Churn Status")
plt.xlabel("Churn")
plt.ylabel("Average Customer Service Calls")
plt.legend(title="Churn Status")
#plt.savefig("task3_barplot_service_calls.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
churn_rate= df_big.groupby("Customer service calls")["Churn"].mean()*100
print(churn_rate)
```

```python
plt.plot(churn_rate.index, churn_rate.values, marker="o", color="tab:red", label="Churn Rate (%)")
plt.title("Churn Rate by Customer Service Calls")
plt.xlabel("Number of customer service calls")
plt.ylabel("Customers who churned (%)")
plt.legend()
# plt.savefig("task3_lineplot_churn_rate.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
sns.scatterplot(x="Total day minutes", y="Total eve minutes", hue="Churn", data=df_big, alpha=0.6)
plt.title("Day Minutes vs Evening Minutes by Churn Status")
plt.xlabel("Total day minutes")
plt.ylabel("Total evening minutes")
plt.legend(title="Churn")
# plt.savefig("task3_scatter_day_vs_eve.png", dpi=300, bbox_inches="tight")
plt.show()
```
